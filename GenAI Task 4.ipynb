```python
paragraph = """
Artificial intelligence has rapidly transformed the way industries operate, 
influencing sectors ranging from healthcare and finance to transportation and 
entertainment. In healthcare, AI-powered systems assist doctors in diagnosing 
diseases earlier and more accurately by analyzing medical images and patient 
data at scale. Financial institutions use machine learning algorithms to detect 
fraudulent transactions, assess credit risk, and automate trading strategies. 
The transportation sector has seen significant advancements through the 
development of self-driving vehicles, which rely on computer vision and sensor 
fusion to navigate complex environments safely. In entertainment, streaming 
platforms leverage recommendation engines powered by AI to personalize content 
for millions of users, increasing engagement and satisfaction. Despite these 
benefits, the rise of AI has also raised concerns about job displacement, data 
privacy, and algorithmic bias. Policymakers and researchers are working together 
to establish ethical guidelines and regulatory frameworks to ensure AI is 
developed and deployed responsibly. Education systems are also adapting, 
introducing AI and data science curricula to prepare the workforce for a 
technology-driven future. As AI models become more sophisticated, questions 
about transparency, accountability, and the potential for artificial general 
intelligence continue to spark debate among experts. Ultimately, the trajectory 
of AI development will depend on collaboration between technologists, 
governments, and society to balance innovation with ethical responsibility, 
ensuring that the benefits of AI are distributed broadly while minimizing 
potential harms.
"""

print("Original word count:", len(paragraph.split()))
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "sshleifer/distilbart-cnn-12-6"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

def summarize(text, max_length=100, min_length=30):
    inputs = tokenizer(text, return_tensors="pt", truncation=True)
    summary_ids = model.generate(
        inputs["input_ids"],
        max_length=max_length,
        min_length=min_length,
        num_beams=4,
        early_stopping=True
    )
    return tokenizer.decode(summary_ids[0], skip_special_tokens=True)

summary_text = summarize(paragraph)
print("Summary:\n", summary_text)
```

```python
original_word_count = len(paragraph.split())
summary_word_count = len(summary_text.split())

print(f"Original word count: {original_word_count}")
print(f"Summary word count: {summary_word_count}")
print(f"Reduction: {round((1 - summary_word_count/original_word_count) * 100, 2)}%")
```
